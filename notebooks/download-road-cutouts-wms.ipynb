## Download Tiles

#### Dependencies

```python
import os

import contextily as ctx
import geopandas as gpd
import matplotlib.pyplot as plt
from owslib.wms import WebMapService

from detect_wildlife_crossings.wms.helpers import download_images_parallel
```

#### Config

```python
depth = 2  # street graph depth
output_dir = f"/media/ruff/raffi-ssd/cassda-data/street_tiles/depth_{depth}"
box_source = f"../data/streets/depth_{depth}"
buffer_size = 640  # meter buffer around each point, square bbox


WMS_URL = "https://image.discomap.eea.europa.eu/arcgis/services/GioLand/VHR_2021_LAEA/ImageServer/WMSServer/?request=GetCapabilities&service=WMS"
CRS = "EPSG:3035"  # ETRS89 / ETRS-LAEA
PIXEL_RESOLUTION_M = 2  # 2 meters per pixel
```

#### Setup

```python
wms = WebMapService(
    WMS_URL,
    version="1.3.0",
)
layer_name = list(wms.contents)[0]
print(layer_name)

countries = gpd.read_file("../data/countries.geojson")
country_codes = countries["ISO3166-1-Alpha-2"]
```

#### Execution

```python
# for every point create a square around it and request a WMS crop centered on the point
img_format = "image/jpeg"
img_size = (
    int(buffer_size * 2 / PIXEL_RESOLUTION_M),
    int(buffer_size * 2 / PIXEL_RESOLUTION_M),
)  # width, height in pixels
```

```python
for i, code in enumerate(country_codes):
    if not os.path.exists(
        f"../data/streets/depth_{depth}/streets_{code}_depth{depth}_boxes.parquet"
    ):
        print(f"Box file for country {code} does not exist, skipping.")
        continue

    print(f"Processing country {code} ({i+1}/{len(country_codes)})")
    gdf = gpd.read_parquet(
        f"../data/streets/depth_{depth}/streets_{code}_depth{depth}_boxes.parquet"
    ).to_crs(CRS)

    gdf.plot()
    plt.title(f"Street boxes for country {code}")
    ctx.add_basemap(
        plt.gca(),
        crs=gdf.crs.to_string(),
        source=ctx.providers.Esri.WorldImagery,
    )
    plt.show()
    
    output_dir_country = os.path.join(output_dir, code)
    os.makedirs(output_dir_country, exist_ok=True)

    # Download with 20 parallel threads
    # Set limit=None to download all, or limit=100 to download first 100, etc.
    results = download_images_parallel(
        gdf=gdf,
        wms=wms,
        layer_name=layer_name,
        img_size=img_size,
        img_format=img_format,
        image_output_dir=output_dir_country,
        crs=CRS,
        max_workers=20,  # Number of parallel threads
        limit=None,  # None = download all, or set a number like 100
    )
```
