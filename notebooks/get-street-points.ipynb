```python
import geopandas as gpd
```

```python
gdf = gpd.read_parquet("data/streets_CH_depth2.parquet")
gdf.crs = "EPSG:4326"
```

```python
gdf = gdf.to_crs(epsg=3035)
```

```python
gdf_filtered = gdf.apply(
    lambda row: row if row["tags"].get("tunnel") != "yes" else None, axis=1
)
```

```python
# buffer 100m around each linestring
gdf_buffered = gdf_filtered.buffer()
```
